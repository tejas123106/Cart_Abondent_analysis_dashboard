```python
session_analysis = df.groupby('user_session').agg(
    user_id=('user_id', 'first'),
    views=('event_type', lambda x: (x == 'view').sum()),
    cart_events=('event_type', lambda x: (x == 'cart').sum()),
    purchase_events=('event_type', lambda x: (x == 'purchase').sum()),
    session_start=('event_time', 'min'),
    session_end=('event_time', 'max')
).reset_index()
```

```python
session_analysis.head()
```

```python
def classify_session(row):
    if row['purchase_events'] > 0:
        return 'Purchased'
    elif row['cart_events'] > 0:
        return 'Abandoned Cart'
    else:
        return 'Browsed Only'

session_analysis['session_status'] = session_analysis.apply(
    classify_session,
    axis=1
)
```

```python
session_analysis['session_status'].value_counts()
```

```python
# Sessions that added something to cart
cart_sessions = session_analysis[
    session_analysis['cart_events'] > 0
]

# Cart sessions that eventually purchased
converted_cart_sessions = cart_sessions[
    cart_sessions['purchase_events'] > 0
]

# Cart sessions with no purchase
abandoned_cart_sessions = cart_sessions[
    cart_sessions['purchase_events'] == 0
]

print("Total Cart Sessions:", len(cart_sessions))
print("Converted Cart Sessions:", len(converted_cart_sessions))
print("Abandoned Cart Sessions:", len(abandoned_cart_sessions))
```

```python
cart_abandonment_rate = (
    len(abandoned_cart_sessions) /
    len(cart_sessions)
) * 100

cart_conversion_rate = (
    len(converted_cart_sessions) /
    len(cart_sessions)
) * 100

print("Cart Abandonment Rate:", round(cart_abandonment_rate, 2), "%")
print("Cart Conversion Rate:", round(cart_conversion_rate, 2), "%")
```

```python
category_analysis = cart_sessions.groupby('category_code').agg(
    cart_sessions=('user_session', 'count'),
    converted_sessions=('purchase_events', lambda x: (x > 0).sum())
).reset_index()

category_analysis['abandoned_sessions'] = (
    category_analysis['cart_sessions']
    - category_analysis['converted_sessions']
)

category_analysis['abandonment_rate'] = (
    category_analysis['abandoned_sessions']
    / category_analysis['cart_sessions']
) * 100

category_analysis = category_analysis.sort_values(
    'abandonment_rate',
    ascending=False
)

category_analysis.head(10)
```

```python
import pandas as pd
import zipfile
import random

zip_path = "data/cart_dataset.zip"

sample_size = 200000
chunk_size = 100000

samples = []

with zipfile.ZipFile(zip_path, "r") as z:
    with z.open("2019-Oct.csv") as f:

        for chunk in pd.read_csv(f, chunksize=chunk_size):

            # Take a random sample from each chunk
            n = min(
                len(chunk),
                sample_size // 50
            )

            samples.append(
                chunk.sample(n=n, random_state=42)
            )

            if sum(len(x) for x in samples) >= sample_size:
                break

df_month = pd.concat(samples, ignore_index=True)

print("Shape:", df_month.shape)
print("Start:", df_month['event_time'].min())
print("End:", df_month['event_time'].max())
```

```python
import pandas as pd
import zipfile

zip_path = "data/cart_dataset.zip"

session_ids = set()

with zipfile.ZipFile(zip_path, "r") as z:
    with z.open("2019-Oct.csv") as f:

        for chunk in pd.read_csv(
            f,
            usecols=['user_session'],
            chunksize=100000
        ):
            session_ids.update(
                chunk['user_session'].dropna().unique()
            )

print("Total unique sessions:", len(session_ids))
```

```python
import random

random.seed(42)

sample_session_ids = random.sample(
    list(session_ids),
    min(20000, len(session_ids))
)

sample_session_ids = set(sample_session_ids)

print("Selected sessions:", len(sample_session_ids))
```

```python
session_sample = []

with zipfile.ZipFile(zip_path, "r") as z:
    with z.open("2019-Oct.csv") as f:

        for chunk in pd.read_csv(f, chunksize=100000):

            selected = chunk[
                chunk['user_session'].isin(sample_session_ids)
            ]

            if len(selected) > 0:
                session_sample.append(selected)

df = pd.concat(
    session_sample,
    ignore_index=True
)

print("Final shape:", df.shape)
print("Unique sessions:", df['user_session'].nunique())
print("Start:", df['event_time'].min())
print("End:", df['event_time'].max())
```

```python
import pandas as pd
import zipfile

zip_path = "data/cart_dataset.zip"

session_sample = []

with zipfile.ZipFile(zip_path, "r") as z:
    with z.open("2019-Oct.csv") as f:

        for chunk in pd.read_csv(f, chunksize=100000):

            selected = chunk[
                chunk['user_session'].isin(sample_session_ids)
            ]

            if not selected.empty:
                session_sample.append(selected)

df = pd.concat(session_sample, ignore_index=True)

print("Reloaded shape:", df.shape)
print("Unique sessions:", df['user_session'].nunique())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("After duplicates:", df.shape)
print("Unique sessions:", df['user_session'].nunique())
```

```python
# Convert event_time to datetime
df['event_time'] = pd.to_datetime(df['event_time'])

# Create session-level summary
session_analysis = df.groupby('user_session').agg(
    user_id=('user_id', 'first'),
    views=('event_type', lambda x: (x == 'view').sum()),
    cart_events=('event_type', lambda x: (x == 'cart').sum()),
    purchase_events=('event_type', lambda x: (x == 'purchase').sum()),
    session_start=('event_time', 'min'),
    session_end=('event_time', 'max')
).reset_index()

print("Total sessions:", len(session_analysis))
print(session_analysis.head())
```

```python
cart_sessions = session_analysis[
    session_analysis['cart_events'] > 0
].copy()

converted_cart_sessions = cart_sessions[
    cart_sessions['purchase_events'] > 0
]

abandoned_cart_sessions = cart_sessions[
    cart_sessions['purchase_events'] == 0
]

print("Total Cart Sessions:", len(cart_sessions))
print("Converted Cart Sessions:", len(converted_cart_sessions))
print("Abandoned Cart Sessions:", len(abandoned_cart_sessions))
```

```python
cart_abandonment_rate = (
    len(abandoned_cart_sessions) / len(cart_sessions)
) * 100

cart_conversion_rate = (
    len(converted_cart_sessions) / len(cart_sessions)
) * 100

print("Cart Abandonment Rate:", round(cart_abandonment_rate, 2), "%")
print("Cart Conversion Rate:", round(cart_conversion_rate, 2), "%")
```

```python
print("===== OVERALL KPIs =====")

print("Total Sessions:", len(session_analysis))

print("Total Users:", df['user_id'].nunique())

print("Total Events:", len(df))

print("Total Views:", (df['event_type'] == 'view').sum())

print("Total Cart Events:", (df['event_type'] == 'cart').sum())

print("Total Purchase Events:", (df['event_type'] == 'purchase').sum())

print("Cart Sessions:", len(cart_sessions))

print("Converted Cart Sessions:", len(converted_cart_sessions))

print("Abandoned Cart Sessions:", len(abandoned_cart_sessions))

print("Cart Abandonment Rate:", round(cart_abandonment_rate, 2), "%")
```

```python
funnel = {
    'Viewed': (session_analysis['views'] > 0).sum(),
    'Added to Cart': (session_analysis['cart_events'] > 0).sum(),
    'Purchased': (session_analysis['purchase_events'] > 0).sum()
}

print(funnel)
```

```python
view_to_cart_rate = (
    funnel['Added to Cart'] / funnel['Viewed']
) * 100

cart_to_purchase_rate = (
    funnel['Purchased'] / funnel['Added to Cart']
) * 100

print("View → Cart Rate:", round(view_to_cart_rate, 2), "%")
print("Cart → Purchase Rate:", round(cart_to_purchase_rate, 2), "%")
```

```python
session_analysis['session_duration_minutes'] = (
    session_analysis['session_end'] -
    session_analysis['session_start']
).dt.total_seconds() / 60

print(session_analysis[['user_session', 'session_duration_minutes']].head())
```

```python
converted_duration = converted_cart_sessions.merge(
    session_analysis[['user_session', 'session_duration_minutes']],
    on='user_session'
)

abandoned_duration = abandoned_cart_sessions.merge(
    session_analysis[['user_session', 'session_duration_minutes']],
    on='user_session'
)

print("Average converted session duration:",
      round(converted_duration['session_duration_minutes'].mean(), 2),
      "minutes")

print("Average abandoned session duration:",
      round(abandoned_duration['session_duration_minutes'].mean(), 2),
      "minutes")
```

```python
converted_views = converted_cart_sessions['views'].mean()
abandoned_views = abandoned_cart_sessions['views'].mean()

print("Average views - Converted:",
      round(converted_views, 2))

print("Average views - Abandoned:",
      round(abandoned_views, 2))
```

```python
print(
    "Average cart events - Converted:",
    round(converted_cart_sessions['cart_events'].mean(), 2)
)

print(
    "Average cart events - Abandoned:",
    round(abandoned_cart_sessions['cart_events'].mean(), 2)
)
```

```python
cart_time = (
    df[df['event_type'] == 'cart']
    .groupby('user_session')['event_time']
    .min()
    .reset_index()
)

cart_time = cart_time.rename(
    columns={'event_time': 'first_cart_time'}
)

cart_time = cart_time.merge(
    cart_sessions[['user_session', 'purchase_events']],
    on='user_session',
    how='left'
)

cart_time['hour'] = cart_time['first_cart_time'].dt.hour

cart_time['status'] = cart_time['purchase_events'].apply(
    lambda x: 'Converted' if x > 0 else 'Abandoned'
)

print(cart_time.head())
```

```python
hourly_analysis = (
    cart_time
    .groupby('hour')
    .agg(
        cart_sessions=('user_session', 'nunique'),
        converted_sessions=('purchase_events', lambda x: (x > 0).sum())
    )
    .reset_index()
)

hourly_analysis['abandoned_sessions'] = (
    hourly_analysis['cart_sessions'] -
    hourly_analysis['converted_sessions']
)

hourly_analysis['abandonment_rate'] = (
    hourly_analysis['abandoned_sessions'] /
    hourly_analysis['cart_sessions'] * 100
)

print(hourly_analysis)
```

```python
hourly_reliable = hourly_analysis[
    hourly_analysis['cart_sessions'] >= 30
].copy()

print(hourly_reliable)
```

```python
print("Highest abandonment among hours with >=30 cart sessions:")
print(
    hourly_reliable.loc[
        hourly_reliable['abandonment_rate'].idxmax()
    ]
)

print("\nLowest abandonment among hours with >=30 cart sessions:")
print(
    hourly_reliable.loc[
        hourly_reliable['abandonment_rate'].idxmin()
    ]
)
```

```python
cart_time['day_of_week'] = cart_time['first_cart_time'].dt.day_name()

daily_analysis = (
    cart_time
    .groupby('day_of_week')
    .agg(
        cart_sessions=('user_session', 'nunique'),
        converted_sessions=('purchase_events', lambda x: (x > 0).sum())
    )
    .reset_index()
)

daily_analysis['abandoned_sessions'] = (
    daily_analysis['cart_sessions'] -
    daily_analysis['converted_sessions']
)

daily_analysis['abandonment_rate'] = (
    daily_analysis['abandoned_sessions'] /
    daily_analysis['cart_sessions'] * 100
)

# Put days in normal Monday → Sunday order
day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

daily_analysis['day_of_week'] = pd.Categorical(
    daily_analysis['day_of_week'],
    categories=day_order,
    ordered=True
)

daily_analysis = daily_analysis.sort_values('day_of_week')

print(daily_analysis)
```

```python
product_data = df[
    df['event_type'] == 'cart'
].copy()

product_data['category_code'] = (
    product_data['category_code']
    .fillna('Unknown')
)

product_data['brand'] = (
    product_data['brand']
    .fillna('Unknown')
)

print("Cart events:", len(product_data))
print("Unique products:", product_data['product_id'].nunique())
```

```python
product_session = product_data[
    [
        'user_session',
        'product_id',
        'category_code',
        'brand',
        'price'
    ]
].drop_duplicates()

product_session = product_session.merge(
    cart_sessions[['user_session', 'purchase_events']],
    on='user_session',
    how='left'
)

product_analysis = (
    product_session
    .groupby(
        ['product_id', 'category_code', 'brand'],
        as_index=False
    )
    .agg(
        cart_sessions=('user_session', 'nunique'),
        converted_sessions=(
            'purchase_events',
            lambda x: (x > 0).sum()
        ),
        average_price=('price', 'mean')
    )
)

product_analysis['abandoned_sessions'] = (
    product_analysis['cart_sessions'] -
    product_analysis['converted_sessions']
)

product_analysis['abandonment_rate'] = (
    product_analysis['abandoned_sessions'] /
    product_analysis['cart_sessions'] * 100
)
```
